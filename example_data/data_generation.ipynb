# Data Generation Notebook

```python
from sklearn.datasets import load_iris
import pandas as pd
```

```python
data = load_iris()
type(data)
```

```python
data.keys()
```

```python
df = pd.DataFrame(data["data"], columns=data["feature_names"])
df["target"] = data["target"]

df.head()
```

```python
df = df.rename(columns={cc: cc.replace("(cm)", "cm").replace(" ", "_") for cc in df.columns})
df.head()
```

```python
df.info()
```

```python
df.to_parquet("iris_dataset.parquet")
```
